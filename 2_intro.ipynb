## Цикл `for`. Операторы `break` и `continue`

Цикл `for` буквально можно воспринимать следующим образом: `для каждого элемента последовательности сделай что-то`. Под последовательностью здесь понимается <b>итерируемый объект</b> т.е. буквально тот, который вы можете растащить на части (например, строку можно раcтащить на символы, список — на элементы списка, кортеж — на элементы кортежа, словарь — на пары ключ-значение). 

````python
years = ['1258', '1658', '1729', '982'] 

for y in years: # для каждого элемента в years
    print(y) # печатаем этот элемент
````

Подумайте: что сейчас хранится в переменной `y`? 

Иногда мы ищем что-то конкретное в списке. Тогда нам можно останоится, как только мы нашли, что искали.  

````python
# break — досрочно выйти из цикла
words = ["собака", "ночь", "инженер", "речь", "фирма"]
# ищем слово из четырёх букв
for word in words: 
    if len(word) == 4:
        print(word)
        break
````

````python
# а теперь ищем все слова длиннее четырёх букв

# continue — пропустить итерацию

for word in words: 
    if len(word) == 4:
        continue # то есть продолжай идти по списку дальше
    else:
        print(word)
````

## Списковые включения

Представим, что нам нужно выбрать только те элементы списка, которые удовлетворяют какому-то условию, и сохранить их к новый список. Для этого надо пройтись по всем элементам списка, проверить, удавлетворяют ли они условию, а затем добавить их в новый список.

Сохраним из списка слов женского рода все слова третьего склонения и укажем им склонение через нижнее подчёркивание.

````python
words = ["собака", "ночь", "шина", "речь", "фирма"]
decl_3 = []
for word in words:
    if word.endswith("ь"): # if word[-1] == "ь":
        decl_3.append(word + "_3")
decl_3
````

````python
# более короткий вариант, который называется "списковые включения"
decl_3 = [word + "_3" for word in words if word.endswith("ь")]
decl_3
````

## Функции `range()` и `enumerate()`. Кортежи
А как изменять элементы списка? Нужно уметь по ним итерироваться и обращаться по индексу к отдельным элементам. То есть нужно знать и индекс, и значение для каждого из элементов в списке при обходе этого спика циклом. Как создать набор индексов, то есть набор чисел от 0 до len(list)? Тут есть два способа: 

1. функция `range()`
2. функция `enumerate()`

`range()` буквально создает диапазон значений от и до (даже может сделать его с определенным шагом). При работе с ней нужно запомнить следующее: 

   `range(START, STOP, STEP)`
   
   + `START` — начало нашего диапазона (дефолтно 0)
   + `STOP` — конец нашего диапазона (не включается!)
   + `STEP` — шаг, с которым мы идем (дефолтно 1)

````python
range(1, 100, 3)
````

````python
# например, сделаем диапазон целых чисел от 1 до 10 с шагом 2 
# то есть возьмем все нечетные числа от 1 до 10
for i in range(1, 11, 2): # 11 снова не включается
    print(i)
````

````python
# значения, которые можно подставить вместо «индексов» можно создать так: 
years = ['1258', '1658', '1729', '982'] 
for i in range(len(years)): 
    print(i)
````

Функция `enumerate()` вернет нам пару индекс-элемент, впредставленную в виде кортежа. 

Кортеж - это упорядоченная неизменяемая коллекция элементов, которую тоже можно индексировать. Он записывается в круглых скобочках. Кстати, если функция возвращает два значения, делает она это в виде кортежа.

````python
for i in enumerate(years):
    print(i)
    print(f'Индексу {i[0]} соответствует элемент {i[1]}') # так можно разбить эту пару
````

Ещё немного про кортежи:

````python
coordinates = (10, 20)
x, y = coordinates  # распаковка
print(x, y)
````

````python
# то есть в цикле выше мы можем распаковать кортеж на две переменные - индекс и значение
years = ['1258', '1658', '1729', '982'] 
for index, value in enumerate(years):
    print(f'Индексу {index} соответствует элемент {value}') # так можно разбить эту пару
````

Давайте поменяем строки на целые числа в нашем списке `years`.

````python
print(years)

for i in range(len(years)):
    years[i] = int(years[i]) 

print(years)
````
