## Лямбда-функции
Бывает такое, что функция потребуется нам всего раз — здесь и сейчас. Тогда зачем нам как-то ее определять и давать ей название? Можно просто это название заменить словом `lambda`.
`lambda`-функции принимают любое количество аргументов, но не могут содержать несколько выражений и всегда возвращают только одно значение.

```python
(lambda var1, var2: var1 ** var2)(2, 8)
```

Самая популярная область применения лямбда-функций — в качестве аргументов в других функциях. В частности лямбда-функции передают в качестве аргументов функциям filter() и map().

```python
my_list = [1, 4, 7]
list(map(lambda x: x ** 2, my_list))
```

```python
numbers = [12, 31, 42, 16]
list(map(lambda x: x if x % 2 == 0 else "uneven", numbers))
```

```python
list(filter(lambda x: x > 0, [-1, 5, 6, -10, 0]))
```

Ещё одна функция высшего порядка, где часто используют лямбда-функции, — это reduce() из модуля functools. Она кумулятивно (с накоплением) применяет заданную функцию к элементам итерируемого объекта и возвращает одно значение.

```python
from functools import reduce
numbers = range(1, 6)
print(reduce(lambda x, y: x + y, numbers))
```

```python
# Соседи, под чьими дверями мяукал Барсик (по порядку обхода)
visits = [
    "Ивановы", "Петровы", "Ивановы", "Сидоровы",
    "Петровы", "Ивановы", "Сидоровы", "Ивановы",
    "Петровы", "Сидоровы", "Ивановы", "Петровы"
]

# Что дали Барсику при каждом визите 
results = [
    "еда", "прогнали", "еда", "погладили",
    "еда", "прогнали", "прогнали", "еда",
    "погладили", "еда", "еда", "прогнали"
]

fed = dict()
chased = dict()
petted = dict()

for i in range(len(visits)):
    if results[i] == "еда":
        if visits[i] not in fed:
            fed[visits[i]] = 0
        fed[visits[i]] += 1
            
    if results[i] == "прогнали":
        if visits[i] not in chased:
            chased[visits[i]] = 0
        chased[visits[i]] += 1
    
    if results[i] == "погладили":
        if visits[i] not in petted:
            petted[visits[i]] = 0
        petted[visits[i]] += 1

print("fed:", fed)
print("chased:", chased)
print("petted:", petted)
```

## Словари

Метод `dict.get(key, default)` - возвращает значение по ключу или возвращает дефолтное значение, если такого ключа нет.

Метод `dict.setdefault(key, default)` - если ключ есть, возвращает его значение. Если ключа нет — создаёт его со значением default и возвращает default.

`collections.defaultdict` — класс, который создаёт словари
с автоматическим значением для отсутствующих ключей. defaultdict удобен, когда мы заранее знаем, какое значение нужно создавать для нового ключа.

```python
products = ["яблоко", "хлеб", "молоко", "яблоко", "сыр", "хлеб", "яблоко"]

counts = {}

for item in products:
    counts[item] = counts.get(item, 0) + 1
```

```python
students = [
    ("Аня", 5), ("Боря", 3), ("Вася", 5), ("Галя", 4), ("Даня", 3),
]

groups = {}

for name, grade in students:
    groups.setdefault(grade, []).append(name)
print(groups)
```

```python
from collections import defaultdict

groups = defaultdict(list)

for name, grade in students:
    groups[grade].append(name)

groups
```

```python
text = '''Лермонтов любил собак. Еще он любил Наталью Николаевну Пушкину.
Только больше всего он любил самого Пушкина.
Читал его стихи и всегда плакал. 
Поплачет, а потом вытащит саблю и давай рубить подушки. 
Тут и любимая собака не попадайся под руку  — штук сорок
как-то зарубил. А Пушкин ни от каких стихов не плакал. Ни за что.'''

def preprocess(text):
    for char in ".,!?—":
        text = text.replace(char, "")
    
    return text.lower().split()


words = preprocess(text)
print(words)
```
