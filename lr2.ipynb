## Лабораторная работа 1. Игра «Быки и коровы»

Создайте консольную игру **«Быки и коровы»**.

В качестве словаря используется файл `summary.txt`. В файле каждое слово записано с новой строки. Перед началом игры необходимо обработать этот файл и получить список слов, которые могут использоваться в игре.

## Правила игры

Компьютер загадывает слово из подготовленного словаря. В словаре хранятся слова длиной 4 или 5 букв, буквы в слова не должны поворяться. Игрок пытается его отгадать. Длина загаданного слова сообщается игроку перед началом игры. Игрок называет слово, пытаясь отгадать загаданное.

Каждая попытка должна удовлетворять тем же требованиям:
* состоять из 4 или 5 букв в зависимости от длины загаданного слова;
* содержать только буквы;
* не содержать повторяющихся букв;
* иметь ту же длину, что и загаданное слово.

При проверке введённого слова программа возвращает количество "быков" и "коров" в слове.

**Бык** — это буква, которая есть в загаданном слове и находится **на правильной позиции**.

Например:

```text
Загадано: МЕТРО
Попытка:  МАСКА
```

Буква `М` находится на первой позиции в обоих словах, поэтому это **1 бык**.


**Корова** — это буква, которая есть в загаданном слове, но находится **на другой позиции**.

Например:

```text
Загадано: МЕТРО
Попытка:  ТЕРКА
```

Буквы, которые присутствуют в обоих словах, но стоят на разных позициях, считаются коровами. После каждой попытки программа должна сообщать:

```text
Быков: 1
Коров: 2
```

При этом программа **не должна сообщать, какие именно буквы являются быками или коровами**.


Игрок продолжает вводить слова до тех пор, пока полностью не угадает загаданное слово.

Программа должна подсчитать количество сделанных попыток. При вводе слова, не удовлетворяюшего условие, количество попыток не увеличивается.

## Подготовка словаря

Прочитайте файл `summary.txt` — это словарь существительных русского языка. Из исходного списка оставьте только слова, которые удовлетворяют **всем** условиям, прописанным в правилах игры. При обработке словаря необходимо использовать **`filter()` или `map()`** и лямбда-функцию.


## Требования к программе

Программа должна быть разбита на **функции**.

Рекомендуется реализовать следующие функции:

`is_valid(word)` проверяет, удовлетворяет ли слова всем правилам игры. Используйте множества для проверки уникальности букв.

`get_bulls(secret, guess)` принимает загаданное слово и попытку и возвращает количество быков. Используйте zip().

`get_cows(secret, guess)` принимает загаданное слово и попытку и возвращает количество коров.

___

Для случайного выбора слова из словаря воспользуйтесь функцией choice из модуля random:

`secret = random.choice(words_for_game)`
