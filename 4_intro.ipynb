````python
# Соседи, под чьими дверями мяукал Барсик (по порядку обхода)
visits = [
    "Ивановы", "Петровы", "Ивановы", "Сидоровы",
    "Петровы", "Ивановы", "Сидоровы", "Ивановы",
    "Петровы", "Сидоровы", "Ивановы", "Петровы"
]

# Что дали Барсику при каждом визите 
results = [
    "еда", "прогнали", "еда", "погладили",
    "еда", "прогнали", "прогнали", "еда",
    "погладили", "еда", "еда", "прогнали"
]
````

````python
fed = dict()
chased = dict()
petted = dict()

for i in range(len(visits)):
    if results[i] == "еда":
        if visits[i] not in fed:
            fed[visits[i]] = 0
        fed[visits[i]] += 1
            
    if results[i] == "прогнали":
        if visits[i] not in chased:
            chased[visits[i]] = 0
        chased[visits[i]] += 1
    
    if results[i] == "погладили":
        if visits[i] not in petted:
            petted[visits[i]] = 0
        petted[visits[i]] += 1

print("fed:", fed)
print("chased:", chased)
print("petted:", petted)
````

````python
list(zip(visits, results))
````

````python
fed = {}
chased = {}
petted = {}

for name, result in zip(visits, results):
    if result == "еда":
        if name not in fed:
            fed[name] = 0
        fed[name] += 1
        
    if result == "прогнали":
        if name not in chased:
            chased[name] = 0
        chased[name] += 1

    if result == "погладили":
        if name not in petted:
            petted[name] = 0
        petted[name] += 1

print("fed:", fed)
print("chased:", chased)
print("petted:", petted)
````

````python
highest_number_fed = max(fed.values())
for a, b in fed.items():
    if b == highest_number_fed:
        print(f"Самые щедрые — {a}. Они покормили кота {highest_number_fed} раза")
````

## Функции

Вы уже познакомились с некоторыми встроенными функциями Python — например, print(), input(), max(), min() и другими. Но часто в программе возникает необходимость повторно использовать один и тот же блок кода. Чтобы не писать его каждый раз заново, программисты выносят такую часть в отдельную функцию.
Функции позволяют:

- переиспользовать один и тот же код без копирования;
- сделать программу более читаемой и структурированной — за счёт разделения логики на небольшие, понятные блоки.

````python
pseudo_kharms = [
    "Однажды Пушкин стрелялся с Гоголем. Пушкин говорит:  «Стреляй первый ты. — Как я? Нет, ты! — Ах, я? Нет ты!»  Так и не стали стреляться.", 
    "Лев Толстой очень любил детей. Приведёт полную комнату, шагу ступить негде, а он все кричит: «Еще! Еще!»",
    "Однажды Гоголь переоделся Пушкиным и пришел к Державину Гавриилу Романовичу. Старик, уверенный, что перед ним и впрямь Пушкин, сходя в гроб, благословил его."
    ]
````

````python
text = pseudo_kharms[0]
text_clean = []

for word in text.split():
    word_clean = word.lower().strip(".,!?«»;:-_—–")
    if word_clean != "":
        text_clean.append(word_clean)
print(text_clean)
````

````python
def clean_text(text):
    text_clean = []
    for word in text.split():
        word_clean = word.lower().strip(".,!?«»;:-_—–")
        if word_clean:
            text_clean.append(word_clean)
    return text_clean
# обратите внимание: НЕ PRINT!!!!!
````

````python
cleaned_pseudo_kharms = []

for text in pseudo_kharms:
    cleaned_pseudo_kharms.append(clean_text(text))

print(cleaned_pseudo_kharms)
````

## Функции высших порядков: filter(), map()

В Python функция может принимать другую функцию в качестве аргумента. Функции, которые принимают другие функции как аргументы, называют функциями высшего порядка.

- map() применяет функцию ко всем элементам коллекции
- filter() отфильтровывает элементы коллекции по условию, прописанному в функции


````python
def only_pos(x):
    return x > 0

list(filter(only_pos, [-1, 5, 6, -10, 0]))
````

````python
my_list = [1, 3, 4, 8]

def square(var1):
    return var1 ** 2

list(map(square, my_list))
````

## Лямбда-функции
Бывает такое, что функция потребуется нам всего раз — здесь и сейчас. Тогда зачем нам как-то ее определять и давать ей название? Можно просто это название заменить словом `lambda`. 
`lambda`-функции принимают любое количество аргументов, но не могут содержать несколько выражений и всегда возвращают только одно значение.

````python
(lambda var1, var2: var1 ** var2)(2, 8)
````

Очень часто лямбда-функции передают в качестве аргументов функциям filter() и map().

````python
my_list = [1, 4, 7]
list(map(lambda x: x ** 2, my_list))
````

````python
numbers = [12, 31, 42, 16]
list(map(lambda x: x if x % 2 == 0 else "uneven", numbers))
````

````python
list(filter(lambda x: x > 0, [-1, 5, 6, -10, 0]))
````
