## Цикл `while`

В языке программирования Python есть два основных типа циклов – циклы `while` и `for`. 

Как выглядит синтаксис цикла `while`?

```
while <условие>:
    <блок кода, который исполняется, пока выполняется логическое условие выше>
```

Идея в том, чтобы цикл не заканчивался до тех пор, пока не будет удовлетворяться какое-то условие. Например, будем запрашивать у пользователя пароль до тех пор, пока он не будет удовлетворять условию.

````python
password = ""

while len(password) != 4:
    password = input("Придумайте пароль (4 символа): ")
    if len(password) < 4:
        print(f"Слишком короткий!")
    if len(password) > 4:
        print(f"Слишком длинный!")

print("Пароль принят!")
````

## Словари и множества

Читаем файлы из рабочей директории.

````python
with open("pizza.txt", "r", encoding="utf-8") as f:
    pizza = f.read()

with open("kebab.txt", "r", encoding="utf-8") as f:
    kebab = f.read()
````

Токенизируем тексты. Пока что не будем приводить слова к начальной форме, просто создадим список лексем в каждом тексте.

````python
import string

words_pizza = []
for token in pizza.split():
    word = token.strip(string.punctuation).lower()
    if word not in string.punctuation:
        words_pizza.append(word)

words_kebab = []
for token in kebab.split():
    word = token.strip(string.punctuation).lower()
    if word not in string.punctuation:
        words_kebab.append(word)

print(words_pizza)
print(words_kebab)
````

Ой, попало тире. Добавим его в пунктуацию.

````python
print(type(string.punctuation))

punct = string.punctuation + "—"
````

````python
words_pizza = []
for token in pizza.split():
    word = token.strip(punct).lower()
    if word:
        words_pizza.append(word)

words_kebab = []
for token in kebab.split():
    word = token.strip(punct).lower()
    if word:
        words_kebab.append(word)

print(words_pizza)
print(words_kebab)
````

Попробуем определить количество уникальных слов в каждом тексте. Для этого нам нужно из списка сделать МНОЖЕСТВО (set). Это изменяемая неупорядоченная коллекция изменяемых типов, которая хранит только уникальные элементы.

````python
unique_p = set(words_pizza)
unique_k = set(words_kebab)

print(f"Всего слов в тексте про пиццу — {len(words_pizza)}, уникальных — {len(unique_p)}.")
print(f"Уникальных слов в тексте про шаурму — {len(words_kebab)}, уникальных — {len(unique_k)}.")
````

Какие операции можно производить над множествами?

````python
print(f'Только в тексте про пиццу: {unique_p - unique_k}')      # разность
print(f'Только в тексте про шаурму: {unique_k - unique_p}')      # разность
print(f'Общие: {unique_p & unique_k}')      # пересечение
print(f'Объединённое множество: {unique_p | unique_k}')     # объединение
````

````python
jaccard = len(unique_p & unique_k) / len(unique_p | unique_k)
print(f"\nСхожесть текстов (Jaccard coeff.): {jaccard:.2f}")
````

Теперь мы хотим посчитать частотность слов в каждом из текстов. Как это сделать? Нам нужно сохранять пару слово — частотность. Для этого могут быть полезны СЛОВАРИ (dict). Это структура данных, которая хранит информацию в виде пар «ключ — значение». Ключом может быть только неизменяемый тип данных, значения могут быть любыми.

````python
# пример
user = {"name": "Алексей", "age": 25}

# доступ к значению по ключу
print(user["name"]) 
print(user["age"]) 
````

````python
# найти ключ по значению
for key, value in user.items():
    if value == "Алексей":
        print(key)
````

````python
pizza_dict = {}
for word in words_pizza:
    if word in pizza_dict:
        pizza_dict[word] += 1
    else: 
        pizza_dict[word] = 1

kebab_dict = {}
for word in words_kebab:
    if word in kebab_dict:
        kebab_dict[word] += 1
    else: 
        kebab_dict[word] = 1

print(pizza_dict)
print(kebab_dict)
````

Сделать частотный словарь из списка можно при помощи специальной функции из модуля collections.

````python
from collections import Counter
pizza_counter = Counter(words_pizza)
kebab_counter = Counter(words_kebab)

print(pizza_counter)
print(kebab_counter)

print("\nТоп-3 слова в оде пицце:")
for word, count in pizza_counter.most_common(3):
    print(f"  {word}: {count}")

print("\nТоп-3 слова в оде шаурме:")
for word, count in kebab_counter.most_common(3):
    print(f"  {word}: {count}")
````
